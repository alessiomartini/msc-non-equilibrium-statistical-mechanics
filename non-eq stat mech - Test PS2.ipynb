we simulate many random walks in 3d and plot them

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
```

```python
def random_walk(dimension, steps): # single walker
    positions = np.zeros((steps + 1, dimension), dtype=int) # una lista 3d: simulazione, step, asse
    returns_to_origin = False # se la simulazione e ripassata dall'origine

    for i in range(steps):
        # Genera direzioni casuali per ogni walker
        moves = np.random.randint(0, dimension) # sceglie la direzione in cui spostarsi
        directions = np.random.choice([-1, 1]) # sceglie se avanti o indietro in quella direzione
        
        # Aggiorna le posizioni
        positions[i+1] = positions[i]
        positions[i+1, moves] += directions
        
        # Controlla se tornano all'origine
        if np.all(positions[i+1] == 0):
            returns_to_origin = True

    return positions, returns_to_origin
```

```python
# Parametri generali
walkers = 1000              # Numero di camminatori
steps = 100              # Numero di passi
dimension = 3            # Numero di dimensioni (3D)

# Simulazione del random walk


# Plot 3D delle traiettorie
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

for i in range(walkers):
    positions, returns_to_origin = random_walk(dimension, steps)
    #print("Walker: ", i, " has passed from the origin: ", returns_to_origin)
    ax.plot(positions[:,0],positions[:,1],positions[:,2])


print("Fraction of passed by the origin: ", np.sum(returns_to_origin)/walkers)

# Configurazione dell'asse
ax.set_title("Traiettorie 3D di 3 Random Walkers")
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")
#ax.legend()
plt.show()
```
